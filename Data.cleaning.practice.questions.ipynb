```python
#19.  Fill missing values in Monthly_Income with the median income.
df["Monthly_Income"]=df["Monthly_Income"].fillna(df["Monthly_Income"].median())
print(df["Monthly_Income"])
print(df["Monthly_Income"].isnull().sum())
```

```python
#*Q20.* Fill missing values in Credit_Score with the median credit score.
df["Credit_Score"]=df["Credit_Score"].fillna(df["Credit_Score"].median())
print(df["Credit_Score"])
print(df["Credit_Score"].isnull().sum())
```

```python
### Module 5: Clean Text Data
#Q36.* Remove extra spaces from the Name column.
df["Name"] = df["Name"].str.strip()

print(df["Name"])
```

```python
#Q37.* Standardize the values in the Gender column so that M, Male, and MALE are represented consistently. Do the same for female values.
# Step 1: Remove extra spaces and standardize capitalization
df["Gender"] = df["Gender"].str.strip().str.title()
# Step 2: Replace short forms with full gender names
df["Gender"] = df["Gender"].replace({"M": "Male","MALE": "Male", "F": "Female", "FEMALE": "Female"})
print(df["Gender"].value_counts(dropna=False))
```

```python
#Q38.* Find inconsistent city names, such as Delhi, New Delhi, and Dehli.
df["City"]=df["City"].str.strip().str.title()
# print(df["City"].value_counts())
df["City"]=df["City"].replace({"Dehli":"Delhi","Bangalore":"Bengaluru","New Delhi":"Delhi","Bombay":"Mumbai"})
print(df["City"].value_counts(dropna=False))
```

```python
#Q47.* Identify unusually high monthly incomes that may be outliers.
#1: IQR (Interquartile Range) method
Q1=df["Monthly_Income"].quantile(0.25)
Q3=df["Monthly_Income"].quantile(0.75)
IQR=Q3-Q1
upper_limit=Q3+1.5*IQR
print("Upper limit:", upper_limit)

print(df[df["Monthly_Income"] > upper_limit])
```

```python
#2: Use a boxplot to visualize outliers
import matplotlib.pyplot as plt
import seaborn as sns
sns.boxplot(data=df,x="Monthly_Income")
plt.title("Monthly Income Outliers")
plt.xlabel("Monthly Income (₹)")
plt.show()
```
